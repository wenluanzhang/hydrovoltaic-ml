```python
SOURCE_FILES = {
    # final processed dataset
    "dataset": ROOT / "notebooks" / "hydrovoltaic_dataset_phase2_input_utf8.csv",

    # Figure 4
    "fig4B_regime_summary": RESULTS / "figure4_resistance_regimes" / "fig4B_R_regime_summary.csv",
    "fig4D_R_descriptor_comparison": RESULTS / "figure4_resistance_regimes" / "fig4D_R_descriptor_comparison.csv",

    # Figure 5
    "fig5A_electrolyte_summary": RESULTS / "figure5_R_origin" / "fig5A_inorganic_electrolyte_summary.csv",
    "fig5B_ion_summary": RESULTS / "figure5_R_origin" / "fig5B_ion_type_summary.csv",
    "fig5C_structure_summary": RESULTS / "figure5_R_origin" / "fig5C_structure_class_summary.csv",
    "fig5_tests": RESULTS / "figure5_R_origin" / "fig5_statistical_tests.csv",

    # Figure 6
    "fig6A_coefficients": RESULTS / "figure6_virtual_design" / "fig6A_linear_descriptor_coefficients_plot.csv",
    "fig6B_model_comparison": RESULTS / "figure6_virtual_design" / "fig6B_linear_vs_polynomial_model_comparison.csv",
    "fig6C_regime_design": RESULTS / "figure6_virtual_design" / "fig6C_regime_aware_virtual_design.csv",
    "fig6D_top_candidates": RESULTS / "figure6_virtual_design" / "fig6D_top_virtual_design_candidates.csv",
}

for name, path in SOURCE_FILES.items():
    print(f"{name:35s}", "OK" if path.exists() else "MISSING", path)
```

```python
rows = []

def add_number(
    section,
    line_hint,
    figure_panel,
    metric,
    group,
    value,
    uncertainty="",
    n="",
    unit="",
    source_file="",
    verified="yes",
    suggested_sentence="",
    notes=""
):
    rows.append({
        "section": section,
        "line_hint": line_hint,
        "figure_panel": figure_panel,
        "metric": metric,
        "group": group,
        "value": value,
        "uncertainty": uncertainty,
        "n": n,
        "unit": unit,
        "source_file": str(source_file),
        "verified": verified,
        "suggested_sentence": suggested_sentence,
        "notes": notes,
    })


def add_check(
    section,
    line_hint,
    figure_panel,
    metric,
    group,
    source_file,
    notes=""
):
    add_number(
        section=section,
        line_hint=line_hint,
        figure_panel=figure_panel,
        metric=metric,
        group=group,
        value="CHECK",
        source_file=source_file,
        verified="check",
        notes=notes,
    )


def read_required_csv(key):
    path = SOURCE_FILES[key]
    if not path.exists():
        raise FileNotFoundError(f"Missing source file for {key}: {path}")
    return pd.read_csv(path, encoding="utf-8-sig")


def fmt(x, digits=3):
    if pd.isna(x):
        return ""
    try:
        return f"{float(x):.{digits}f}"
    except Exception:
        return str(x)
```

```python
for label, col in COLS.items():
    if col not in df.columns:
        add_check(
            section="2.1/2.2",
            line_hint="157-195",
            figure_panel="Fig. 1D",
            metric=f"{label}_availability",
            group=label,
            source_file=SOURCE_FILES["dataset"],
            notes=f"Column not found: {col}"
        )
    else:
        count = int(df[col].notna().sum())
        add_number(
            section="2.1/2.2",
            line_hint="157-195",
            figure_panel="Fig. 1D",
            metric=f"{label}_availability",
            group=label,
            value=count,
            n=N_TOTAL,
            unit=f"records out of {N_TOTAL}",
            source_file=SOURCE_FILES["dataset"],
        )
```

```python
# Section 2 descriptor counts
# Structure counts
structure_col = COLS["structure"]
if structure_col in df.columns:
    structure_counts = df[structure_col].value_counts(dropna=False).reset_index()
    structure_counts.columns = ["group", "count"]
    structure_counts.to_csv(OUT_DIR / "section2_structure_counts.csv", index=False, encoding="utf-8-sig")
    
    for _, r in structure_counts.iterrows():
        add_number(
            section="2.3",
            line_hint="224-226",
            figure_panel="Fig. 1C",
            metric="structure_count",
            group=str(r["group"]),
            value=int(r["count"]),
            n=N_TOTAL,
            unit="records",
            source_file=OUT_DIR / "section2_structure_counts.csv",
        )

# Ion counts
ion_col = COLS["ion"]
if ion_col in df.columns:
    ion_counts = df[ion_col].value_counts(dropna=False).reset_index()
    ion_counts.columns = ["group", "count"]
    ion_counts.to_csv(OUT_DIR / "section2_ion_counts.csv", index=False, encoding="utf-8-sig")
    
    for _, r in ion_counts.iterrows():
        add_number(
            section="2.3",
            line_hint="224-226",
            figure_panel="Fig. 1C",
            metric="ion_count",
            group=str(r["group"]),
            value=int(r["count"]),
            n=N_TOTAL,
            unit="records",
            source_file=OUT_DIR / "section2_ion_counts.csv",
        )
```

```python
# Section 3 group summaries 这里从最终 dataset 重新算 Fig. 2A–C 的 group count 和 median。这不是读图，是从数据算。
target_col = COLS["target"]

def add_group_summary(section, line_hint, figure_panel, group_col, metric_name, out_name):
    if group_col not in df.columns:
        add_check(
            section=section,
            line_hint=line_hint,
            figure_panel=figure_panel,
            metric=metric_name,
            group="all",
            source_file=SOURCE_FILES["dataset"],
            notes=f"Missing group column: {group_col}"
        )
        return
    
    if target_col not in df.columns:
        add_check(
            section=section,
            line_hint=line_hint,
            figure_panel=figure_panel,
            metric=metric_name,
            group="all",
            source_file=SOURCE_FILES["dataset"],
            notes=f"Missing target column: {target_col}"
        )
        return
    
    d = df[[group_col, target_col]].copy()
    d[target_col] = pd.to_numeric(d[target_col], errors="coerce")
    d = d.dropna(subset=[group_col, target_col])
    
    summary = (
        d.groupby(group_col)[target_col]
        .agg(["count", "mean", "median", "std", "min", "max"])
        .reset_index()
    )
    
    out_path = OUT_DIR / out_name
    summary.to_csv(out_path, index=False, encoding="utf-8-sig")
    
    for _, r in summary.iterrows():
        add_number(
            section=section,
            line_hint=line_hint,
            figure_panel=figure_panel,
            metric=metric_name,
            group=str(r[group_col]),
            value=fmt(r["median"], 3),
            uncertainty=f"mean={fmt(r['mean'],3)}, std={fmt(r['std'],3)}",
            n=int(r["count"]),
            unit="log(P_est)",
            source_file=out_path,
        )

add_group_summary(
    section="3.1",
    line_hint="257-263",
    figure_panel="Fig. 2A",
    group_col=COLS["mechanism"],
    metric_name="mechanism_group_median_log_P_est",
    out_name="section3_mechanism_group_summary.csv"
)

add_group_summary(
    section="3.2",
    line_hint="271-278",
    figure_panel="Fig. 2B",
    group_col=COLS["structure"],
    metric_name="structure_group_median_log_P_est",
    out_name="section3_structure_group_summary.csv"
)

add_group_summary(
    section="3.2",
    line_hint="279-285",
    figure_panel="Fig. 2C",
    group_col=COLS["ion"],
    metric_name="ion_group_median_log_P_est",
    out_name="section3_ion_group_summary.csv"
)
```
