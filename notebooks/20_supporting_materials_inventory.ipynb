```python
def status_from_notes(text, expected_files_df):
    """
    Lightweight status helper.
    This does not fully validate scientific content;
    it only checks whether likely result files exist.
    """
    text = str(text).lower()

    if "manual table" in text:
        return "manual_creation_needed"

    keywords = []
    if "fig1" in text:
        keywords.append("figure1_dataset_overview")
    if "fig3" in text:
        keywords.append("figure3_model_comparison")
    if "fig4" in text:
        keywords.append("figure4_resistance_regimes")
    if "fig5" in text:
        keywords.append("figure5_R_origin")
    if "fig6" in text:
        keywords.append("figure6_virtual_design")

    if not keywords:
        return "source_check_needed"

    mask = expected_files_df["relative_path"].str.contains("|".join(keywords), case=False, regex=True)
    subset = expected_files_df.loc[mask]

    if len(subset) == 0:
        return "source_check_needed"
    elif subset["exists"].all():
        return "source_files_found"
    elif subset["exists"].any():
        return "partial_source_files_found"
    else:
        return "source_files_missing"


si_tables_df["source_file_status"] = si_tables_df["expected_source_files"].apply(
    lambda x: status_from_notes(x, expected_files_df)
)

si_figures_df["source_file_status"] = si_figures_df["expected_source_files"].apply(
    lambda x: status_from_notes(x, expected_files_df)
)

display(si_tables_df[["item", "title", "priority", "source_file_status", "notes"]])
display(si_figures_df[["item", "title", "priority", "source_file_status", "notes"]])
```
